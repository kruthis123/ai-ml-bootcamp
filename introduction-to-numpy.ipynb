## Dot product

```python
mat1 = np.random.randint(10, size=(5, 3))
mat2 = np.random.randint(10, size=(5, 3))
mat1, mat2
```

```python
# Elementwise multiplication (Hadamard product)
mat1 * mat2
```

```python
np.dot(mat1, mat2.T)
```
